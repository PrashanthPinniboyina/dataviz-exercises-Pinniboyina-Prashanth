```python
import pandas as pd
import plotly.express as px


# Dataset: Gapminder — GDP, Life Expectancy, Population by Country
# Source: Gapminder Foundation (gapminder.org)

df = px.data.gapminder()
print(f"Loaded: {len(df)} rows")
print(df.head())
```

```python
# explore

print(df.info())
print("Years:", sorted(df['year'].unique()))
print("Continents:", df['continent'].unique())
print(df.describe().round(1))
```

## Task 1 — Scatter: life expectancy change over time

**What to build:** A scatter showing **GDP per capita vs life expectancy** for **two years** (2002 and 2007) to show how both moved — use **colour for year** (just 2 colours), **one continent only**.

Choose any continent except Africa (that was the example). Highlight the change direction.

> 💡 Filter: `df.loc[df['continent'] == 'YOUR_CHOICE']` then filter years


```python
import plotly.express as px
import plotly.graph_objects as go

# Choose one continent
continent_choice = 'Asia'

# Filter data for selected continent and years
filtered = df[
    (df['continent'] == continent_choice) &
    (df['year'].isin([2002, 2007]))
].copy()

# Create scatter plot
fig = px.scatter(
    filtered,
    x='gdpPercap',
    y='lifeExp',
    color=filtered['year'].astype(str),
    hover_name='country',
    color_discrete_map={
        '2002': '#3498db',
        '2007': '#e74c3c'
    },
    title='Asia: GDP per Capita and Life Expectancy Improved from 2002 to 2007'
)

# Add lines showing direction of change
for country in filtered['country'].unique():
    country_data = filtered[
        filtered['country'] == country
    ].sort_values('year')

    if len(country_data) == 2:
        fig.add_trace(
            go.Scatter(
                x=country_data['gdpPercap'],
                y=country_data['lifeExp'],
                mode='lines',
                line=dict(color='lightgrey', width=1),
                showlegend=False,
                hoverinfo='skip'
            )
        )

fig.update_layout(
    xaxis_title='GDP per Capita',
    yaxis_title='Life Expectancy (Years)',
    template='plotly_white'
)

fig.show()
```

## Task 2 — Bubble chart: tell a story

**What to build:** A bubble chart (full 2007 dataset, all countries) where:
- x = GDP per capita (log scale)
- y = life expectancy
- size = population
- colour = ONE continent highlighted (your choice), all others grey
- At least one annotation explaining the highlighted group's story

> This is the grey-and-highlight technique applied to a bubble chart.


```python
import plotly.express as px

# Filter data for 2007
df_2007 = df[df['year'] == 2007].copy()

# Highlight Asia, show other continents in grey
df_2007['Group'] = df_2007['continent'].apply(
    lambda x: 'Asia' if x == 'Asia' else 'Other Continents'
)

# Create bubble chart
fig = px.scatter(
    df_2007,
    x='gdpPercap',
    y='lifeExp',
    size='pop',
    color='Group',
    hover_name='country',
    color_discrete_map={
        'Asia': '#E74C3C',
        'Other Continents': '#D3D3D3'
    },
    size_max=50,
    log_x=True,
    title='Asia Shows Diverse Life Expectancy and Economic Growth in 2007'
)

# Add annotation
fig.add_annotation(
    x=1500,
    y=42,
    text="Asia: Wide differences in income<br>and life expectancy",
    showarrow=True,
    arrowhead=2,
    ax=80,
    ay=-55,
    bgcolor="white",
    bordercolor="#E74C3C",
    borderwidth=1,
    font=dict(size=11, color="black")
)

fig.update_layout(
    xaxis_title='GDP per Capita (Log Scale)',
    yaxis_title='Life Expectancy (Years)',
    template='plotly_white'
)
fig.add_annotation(
    x=0.02,
    y=0.95,
    xref='paper',
    yref='paper',
    text="Asia shows wide economic differences, but wealthier countries generally have higher life expectancy.",
    showarrow=False,
    bgcolor='lightyellow',
    bordercolor='gray',
    borderwidth=1,
    font=dict(size=11, color='black'),
    align='left'
)

fig.show()
```
